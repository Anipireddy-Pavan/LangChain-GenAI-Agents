```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

model = ChatGroq(
    model="llama-3.1-8b-instant"
)

response = model.invoke("Could you please explain what AI is?")

print(response.content)
```

```python
#Streaming

"""
Most models can stream their output content while it is being generated.
By displaying output progressively , streaming significantly improver user experience, particularly for longer responses.
Calling stream() returns an iterator that yields chunks as they are produced.
You can loop to process each chunk in real time.
"""
```

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence?")
```

```python
for chunk in model.stream("Write a 200-word paragraph about Artificial Intelligence."):
    print(chunk.content, end="", flush=True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="|", flush=True)
```

```python
#Batch.
"""
Batching a collection of independent requests to a model can significantly improve performence and reduce the costs, as the processing can be done in paralle.
"""
```

```python
responses = model.batch([
    "Why do parrots have colorful features",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(responses)
```

```python
model.batch(
    [
      "Why do parrots have colorful features",
      "How do airplanes fly",
      "What is quantam computing?"  
    ],
    config = {
        "max_concurrency": 5, #limit to 5 parallel calls
    }
)
```
